```python
#loop through all the objects column and encode them
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
for col in fd_encoded.select_dtypes(include=['object']).columns:
    fd_encoded[col] = label_encoder.fit_transform(fd_encoded[col].astype(str))

fd_encoded.dtypes.head(20)
```

```python
keep_cols = [
    "Age", "Gender", "Height (m)", "Weight (kg)",
    "meal_type", "diet_type", "Workout_Type", "Calories_Burned", "Workout_Frequency (days/week)",
    "cooking_method", "prep_time_min", "cook_time_min", "rating", "Physical exercise", "diet_type",
    "Proteins", "Carbs", "Fats", "sugar_g", "Calories", "Name of Exercise",
    "sodium_mg", "cholesterol_mg", "serving_size_g"
]

new_final_data = fd_encoded[keep_cols]
```

```python
# Health Score = Proteins - (Fats + Sugar)
# Step 1: Calculate Health Score
new_final_data['Health_Score'] = new_final_data['Proteins'] - (new_final_data['Fats'] + new_final_data['sugar_g'])

# Step 2: Label meals as Healthy or Unhealthy
new_final_data['Health_Label'] = new_final_data['Health_Score'].apply(lambda x: 'Healthy' if x > 0 else 'Unhealthy')
```
