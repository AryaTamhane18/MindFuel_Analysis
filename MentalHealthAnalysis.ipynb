Mental Health Dataset

```python
# Importing the libraries

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Loading the dataset 

df = pd.read_csv("Mental Health Dataset.csv")
```

```python
# Checking number of rows and columns are present in the dataset

print("🔹 Dataset Shape (rows, columns):", df.shape)
```

```python
# Seeing the overview of the data 

df.head(10)
```

```python
# Finding out the missing values

df.isnull().sum()
```

```python
# Finding the most common values from the Self_Employed column as it has missing values

''' Since the Self employed column has only 0.18% missing values I have replaced it with the most common value. As 
removing the rows could cause loss of specific useful data.'''

most_common_value = df['self_employed'].mode()[0]
print("Most common value in 'self_employed':", most_common_value)
```

```python
# Replacing all the missing rows with the most important value "NO"

df['self_employed'].fillna(most_common_value, inplace=True)
```

```python
print("\nMissing values after filling:")
print(df.isnull().sum())
```

```python
# Check for duplicates 

duplicates_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates_count}")
```

```python
# Display the first few duplicate rows
df[df.duplicated()].head()
```

```python
# Removing the duplicates

df = df.drop_duplicates()
```

```python
print("Remaining duplicates:", df.duplicated().sum())
```

```python
# Checking number of rows and columns are present in the dataset after cleaning 

print("Dataset Shape (rows, columns):", df.shape)
```

```python
# Checking the datatypes for the columns 

df.dtypes
```

```python
# Drop the 'Timestamp' column as it is not present in my other dataset 

df = df.drop(columns=['Timestamp'])
```

```python
 print(df.head())
df.shape
```

```python
# Calculating the 'Stress Score' column

# Grouping the columns with stress relation

stress_cols = ['Growing_Stress', 'Coping_Struggles', 'Mood_Swings', 'Changes_Habits']
```

```python
# Checking the unique values in each columns 

for col in stress_cols:
    print(f"{col} unique values: {df[col].unique()}")
```

```python
# Maping these categorical data into numerical values 

# Define mapping dictionaries 
growing_map = {'No': 0, 'Maybe': 0.5, 'Yes': 1}
coping_map = {'No': 0, 'Yes': 1}
mood_map = {'Low': 0, 'Medium': 0.5, 'High': 1}
changes_map = {'No': 0, 'Maybe': 0.5, 'Yes': 1}

# Apply the mapping
df['Growing_Stress'] = df['Growing_Stress'].map(growing_map)
df['Coping_Struggles'] = df['Coping_Struggles'].map(coping_map)
df['Mood_Swings'] = df['Mood_Swings'].map(mood_map)
df['Changes_Habits'] = df['Changes_Habits'].map(changes_map)

```

```python
df.head()
```

```python
# Taking the average of the stress score columns

stress_cols = ['Growing_Stress', 'Coping_Struggles', 'Mood_Swings', 'Changes_Habits']

df['Stress_Score'] = df[stress_cols].mean(axis=1)
```

```python
df.head()
```

```python
# Classifying the Stress Score as Healthy and Unhealthy 

df['Lifestyle_Category'] = df['Stress_Score'].apply(
    lambda x: 'Healthy' if x < 0.5 else 'Unhealthy'
)
```

```python
df.head()
```

```python
# Select only numeric columns from your mental health dataframe
numeric_df = df.select_dtypes(include=['int64', 'float64'])

# Check what numeric columns were selected
print(numeric_df.columns)
```

```python
corr = numeric_df.corr()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
plt.imshow(corr, cmap='viridis')
plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha='right')
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Heatmap of All Numeric Columns")
plt.tight_layout()
plt.show()
```

```python
# Identifying the numerical columns 

numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
print(numerical_cols)
```

```python
# Creating histograms 

for col in numerical_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=30)
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()

'''
Stress score ranges from 0 to 1,:
--Most values cluster between 0.3 and 0.7
--Very few values are near 0 or 1
--The distribution looks normal (bell-shaped)
--Most respondents have a moderate stress score
-- Very low and very high scores are less common
'''
```

```python
for col in numerical_cols:
    plt.figure(figsize=(6,4))
    plt.boxplot(df[col])
    plt.title(f"Boxplot of {col}")
    plt.ylabel(col)
    plt.show()


'''
--All stress score values fall within the expected range
--There are no extreme or abnormal entries
--No data cleaning (removal or correction of outliers) is required for this column    
--The central box (IQR) lies between approximately 0.38 and 0.62, 
meaning that 50% of the respondents have moderate stress levels.
'''
```
